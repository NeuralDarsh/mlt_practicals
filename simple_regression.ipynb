```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data ={"hours" : [1,2,3,4,5],
       "marks" : [50,60,80,90,100]}
df=pd.DataFrame(data)
print(df)
```

```python
X=df[["hours"]].values
y=df["marks"].values
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,y)
```

```python
slope=model.coef_[0]
intercept=model.intercept_
```

```python
print(f"slope(m):{slope}")
print(f"intercept (b):{intercept}")
print(f"Equation of the regression line: y={slope:.2f} x + {intercept:.2f}")
```

```python
df["Predicted Marks"]=model.predict(X)
```

```python
df.head()
```

```python
plt.scatter(df['hours'],df['marks'],color='blue',label='Actual Marks')
```

```python
plt.plot(df['hours'],df['Predicted Marks'],color='red',label='Regression Line')
plt.xlabel('Hours Studied')
plt.ylabel('marks')
plt.title('Hours vs Marks')
plt.legend()
plt.show()
```
